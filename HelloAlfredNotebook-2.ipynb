# AI-Powered Behavioral Nudging in Atrial Fibrillation (AF) Care

# **Business Case and Background Story**

Atrial Fibrillation (AF) is a prevalent cardiac condition that increases stroke risk, requiring patients to adhere to medications (like anticoagulants) and lifestyle adjustments. However, maintaining these behaviors long-term is challenging – nonadherence to chronic cardiovascular medications has been linked to higher rates of hospitalization and mortality. This presents a clear need and business case for interventions that improve patient adherence and engagement in AF care.

Digital health solutions with AI-powered behavioral nudges offer a promising approach. By leveraging smartphones, wearables, and patient data, we can deliver personalized “nudges” – subtle prompts or suggestions – that guide patients toward healthier actions without overt coercion. Early studies indicate that tailored messaging interventions for AF patients can improve patient knowledge, engagement, and adherence to therapy. For healthcare providers and payers, better adherence translates to fewer complications and hospital visits, improving outcomes while reducing costs.

Behavioral science provides a strong foundation for designing these nudges. Frameworks like MINDSPACE (Messenger, Incentives, Norms, Defaults, Salience, Priming, Affect, Commitments, Ego) enumerate key psychological levers to influence behavior. Originally developed for public policy, MINDSPACE has been widely adopted in health contexts to craft effective interventions. Complementing this is the EAST framework (Easy, Attractive, Social, Timely), which serves as a simple checklist for nudge design: make the desired behavior easy to do, present it attractively, leverage social influences, and time it well. These principles ensure that our AI-driven nudges are grounded in what has worked in behavioral economics and psychology – for example, using a credible messenger (like a doctor persona) to deliver advice, or setting the default option to the healthy choice.

Another cornerstone is understanding patient readiness to change. The Transtheoretical Model (TTM), developed by Prochaska and DiClemente, views behavior change as a progression through stages: Precontemplation (not ready), Contemplation (thinking about change), Preparation (getting ready), Action (actively making changes), and Maintenance (sustaining change). TTM emphasizes that people move back and forth through these stages, rather than instantly changing behavior. In fact, TTM has been called “the dominant model of health behavior change” due to its broad application across behaviors. Researchers (e.g. UMBC’s HABITS Lab led by TTM’s co-developer) stress tailoring interventions to a person’s current stage, rather than using a one-size-fits-all approach. For our project, this means the background story of each patient matters – an AF patient in a Precontemplation stage (unaware or not convinced of the need for change) will need different support than one in Preparation or Action.

We also integrate the COM-B model (Capability, Opportunity, Motivation – Behavior), a widely used behavior change framework for diagnosing why a behavior is not occurring. COM-B posits that a person needs the capability (knowledge/skills and physical ability), the opportunity (social and environmental supports), and the motivation (desire and habits) to perform a behavior. If any element is lacking, it becomes a barrier. For example, an AF patient might have low capability (poor understanding of their condition), or limited opportunity (no routine or tools to remember meds), or low motivation (perhaps they don’t feel symptoms, so they forget their medication). Identifying these barriers helps target the intervention appropriately.

# **Problem Statement and Top Three Challenges**

**Problem Statement:** AF patients often struggle with continuous self-management. Despite knowing the stakes (avoiding strokes, managing symptoms), many patients falter in taking medications on time, monitoring their vitals, or maintaining lifestyle changes. This gap between knowing and doing can be attributed to various personal and systemic barriers. Traditional interventions like clinic education or generic reminder texts are limited – they are not tailored to individual patient needs and often result in alert fatigue (patients start ignoring the barrage of generic reminders). In short, one-size-fits-all approaches fail to consistently change behavior in a diverse patient population. We need a smarter system that personalizes support for each patient’s situation in real time.

**Top Three Challenges:** In designing an AI-powered behavioral nudging solution for AF care, we have identified three major challenges:

**Personalization to Individual Readiness and Barriers:** Every patient is different – one might forget doses due to a busy schedule, another might be in denial about their condition, and yet another might lack knowledge. A core challenge is accurately assessing each patient’s behavioral readiness and barriers, then tailoring interventions accordingly. Both TTM and COM-B underscore this need for personalization: people are at different stages and have different limiting factors, so the intervention must meet them where they are. Implementing this is non-trivial; it requires gathering the right data (e.g. patient knowledge level, routines, motivations) and possibly segmenting patients. For example, research shows AF patients can be clustered into distinct subgroups (“stressed and discontented,” “complacent,” etc.), each with different challenges and support needs. The challenge is to build a system that can identify these individual or subgroup differences and avoid a blanket approach.

**Designing Effective and Engaging Nudges (Behavioral Design Challenge):** Even if we know what a patient’s barrier is, how do we nudge them effectively? Crafting a message or prompt that actually changes behavior is an art and science. We must incorporate behavioral principles (MINDSPACE/EAST) into the content and delivery of the nudge. The difficulty lies in operationalizing these frameworks: ensuring each nudge is Easy to act on, has an Attractive presentation, leverages Social norms appropriately, and is Timely. For instance, using incentives or social proof can boost motivation, but overdoing it may backfire; using an authoritative Messenger (like a note from “Dr. Smith”) can add credibility, but only if the patient trusts that source. We also have to maintain a supportive tone (avoiding guilt or fear) and respect patient autonomy. Essentially, there’s a UX design challenge in translating psychology into a chatbot or app interface. All nudges should be helpful, not annoying – this requires careful content creation and perhaps A/B testing of different nudge designs.

**Timeliness and Context Awareness (Technical Integration Challenge):** Timing is everything for behavior change. A perfectly crafted reminder is useless if it comes at the wrong moment (e.g., a reminder to exercise when the patient is at work meeting). Delivering Just-In-Time Adaptive Interventions (JITAI) is technically challenging because it demands real-time data and context integration. Our system needs to detect context from various data streams – medication schedules, wearable sensors (heart rate, activity), maybe location or time of day – to send nudges at the optimal moment. For example, if a smartwatch or smart pill bottle indicates a dose was missed, the system should trigger a prompt immediately. Or if a patient’s stress level is rising (detected via a wearable or self-report), it might be timely to send a breathing exercise suggestion. Implementing this requires robust data ingestion from devices/EHR, event-triggered messaging, and handling asynchronous conditions (network delays, data inaccuracies). We also must avoid overwhelming the user – if our algorithm is too sensitive, it could send too many alerts (leading to alert fatigue). In summary, the technical challenge is building an integrated pipeline that continuously senses, decides, and acts, all while balancing frequency and timing to maximize receptivity.

Additional challenges include ensuring patient privacy and data security, gaining user trust (especially with AI making suggestions – transparency is key), and measuring long-term effectiveness (since behavior change outcomes like stroke reduction may take time to observe). Moreover, demonstrating ROI to stakeholders (hospitals, insurers) will require solid evidence that these nudges improve adherence in a cost-effective way. We keep these in mind, but the three challenges above are the primary focus as we design the solution.

2. **Behavioral Readiness Assessment:** Once we have the patient’s data, the next step is to assess their behavioral profile – essentially diagnosing why they might struggle with the target health behaviors. We use the COM-B model here to evaluate three dimensions:

    Capability: Does the patient have the ability/knowledge to perform the behavior? (e.g., health literacy, physical ability)

    Opportunity: Do they have the opportunity and resources? (e.g., a stable routine, family support, tools like pillboxes)

    Motivation: How motivated and habitually inclined are they? (e.g., their confidence, past adherence patterns)

    Our system will analyze the patient data to assign a status for each of these (perhaps high/medium/low or a numeric score). For instance, we might administer a brief quiz to gauge AF knowledge (capability), use wearable data to see if they have a daily routine (opportunity), and check app interaction logs for engagement (motivation). If sufficient data is available, a machine learning classifier could be trained to predict these labels (using patterns from past patients), although our pseudocode uses rule-based evaluation for simplicity. We also consider integrating Transtheoretical Model (TTM) insights here: for example, mapping a COM-B combination to a TTM stage (low motivation + low capability might indicate the patient is in a “precontemplation” or “contemplation” stage regarding the behavior). This helps ensure we capture not just what the patient is struggling with, but also how ready they are to change.

If we know the cluster, we gain additional context – e.g., patients in the “stressed” cluster might overall need stress-management support. Thus, cluster membership can become an input to the next step, ensuring our strategy selection is informed by both individual traits and group trends.

3. **Identify Target Behavior and Key Barrier:** With the patient’s state assessed, we pinpoint what behavior to focus on and what barrier is most important. The target behavior could be one of several health behaviors relevant to AF care – common examples: “medication adherence (take medications on schedule)”, “physical activity (daily walk or exercise)”, “self-monitoring (e.g., take pulse or blood pressure daily)”, or “appointment follow-ups”. We determine this target based on clinical priority and the patient’s data. For instance, if our patient has good medication adherence but very low activity levels, the system might choose to target exercise habit formation. In our pseudocode, determine_target_behavior(patient) would use rules or physician input to select this.

    Next, we identify the key barrier preventing the target behavior. This usually ties back to the weakest link in the COM-B profile. For example, if the patient’s profile was capability=low (perhaps they don’t fully understand their medication regimen), that suggests the barrier is a knowledge gap (capability). If motivation was marked low (maybe they often forget or don’t see the importance), then low motivation is the barrier. In simple terms, we apply logic like: if capability is the lowest dimension, focus on that; else if motivation is lowest, focus on motivation, etc.. This picks one primary barrier to address first. (Realistically, multiple factors often overlap, but for intervention design it helps to target one at a time or prioritize.) This step is crucial because it translates our assessment into a concrete problem statement for the AI coach: e.g., “The patient needs help with motivation to walk daily” or “The patient needs education about AF to improve their capability.” We could implement this logic as a classifier or simple decision tree – indeed, this is a point where a ML classifier could live in the pipeline (predicting the barrier category from patient features, instead of using hard-coded rules).

4. **Nudge Strategy Selection (Using Behavioral Frameworks):** Given the target behavior and primary barrier, our system next decides what type of nudge or intervention to deploy. We have a toolbox of nudge strategies categorized by the kind of problem they solve:

    If the main barrier is capability (knowledge/skill deficit) – the strategy might be an Educational Nudge (providing information, tutorials, feedback to build ability).

    If the barrier is opportunity (forgetfulness or environmental obstacles) – the strategy might be an Environmental Prompt or Reminder (i.e., prompt the patient at the right time, or help restructure the environment, like setting up a pillbox or phone alarm).

    If the barrier is motivation (low drive or confidence) – the strategy might be a Motivational Nudge (leveraging encouragement, incentives, social proof, commitment devices, etc.).

    Our pseudocode function select_nudge_strategy(target_behavior, barrier) implements a simple version of this mapping. For example, if barrier == "motivation", it sets strategy = "MotivationalNudge". This strategy is essentially a high-level plan (not the exact message, but the approach we will take).

    Critically, we do not select the strategy in a vacuum – we incorporate the MINDSPACE and EAST principles to refine it. This means once we choose the strategy type, we ask: How can we make this nudge most effective according to behavioral science? For instance, suppose the strategy is a motivational nudge to improve medication adherence. We might decide to apply the Messenger principle by framing the reminder as coming from a trusted clinician, use an Incentive by emphasizing the immediate benefit (“taking your pill today keeps your stroke risk low – that’s a win for you”), and add Salience by highlighting the action (“Take your dose now”) clearly. At the same time, we check it against EAST: make sure it’s Easy (one-click confirmation of dose taken in the app), Attractive (maybe include a friendly emoji or progress badge for motivation), Social (maybe mention “most patients like you took theirs today” if appropriate), and Timely (send exactly when needed). We essentially design the nudge on paper in this step – defining its content outline, tone, and trigger conditions according to these principles. The outcome is a nudge_plan that might look like: {"type": "MotivationalNudge", "framework_tactics": ["SocialProof", "Timely", "Messenger=Doctor"]} to guide the content creation. (In code, this was represented as strategy_design after applying frameworks.)

5. **Personalized Content Selection:** With a strategy in hand, we now generate the actual content of the nudge that the patient will see. This is where personalization truly comes to life. We either select a pre-written message template or dynamically generate text, but we tailor it to the individual patient’s context. Our pseudocode function personalize_nudge_content(nudge_plan, patient) handles this. Key considerations in this step:

Taking your anticoagulant reduces AF-related stroke risk by 60%”). For a Reminder/Prompt, the content is a direct reminder message (e.g., “Hi John, it’s 8:00 PM – time to take your heart medication.”). For a Motivational nudge, the content might be a encouraging message or a quiz that engages the patient.

    - We use patient-specific details in the content. This can include the patient’s name, their specific medication or behavior, or reference something from their profile.     - The content is phrased in a user-friendly way, following the tone guidelines (supportive, empowering, non-judgmental). For instance, to improve opportunity (environment), we might send: “It’s 8 AM – time to take your anticoagulant. I’ve logged it for you with one tap!” which makes it simple and reminds them in context. To address motivation, we might say: “Keep it up! Taking your medication now keeps you on track – you’re doing great 👍. (Did you know 9 out of 10 AF patients at our clinic take theirs on time?)”, combining encouragement and a norm reference.

    - If we have multiple channels (app, SMS, email), we also ensure the content fits the medium (short text for SMS vs. richer content in app). Our function might choose different templates based on patient.preferred_channel.

    The result of this step is a fully personalized nudge message ready to be delivered, encapsulated as nudge_content. Importantly, this step is where we differentiate ourselves from generic messaging – by removing this personalization step, a system would default to generic one-size-fits-all content, which is far less effective. 6. **Scheduling the Nudge (Just-In-Time Adaptive Delivery):** Deciding when to deliver the nudge is as important as the content. We employ JITAI logic to schedule the nudge optimally. In our pseudocode, schedule_nudge_delivery(patient, content, nudge_plan) encapsulates this. There are a few scenarios:

    - If the nudge is tied to a specific event or condition, we trigger it immediately. For example, if our data indicates the patient just missed a medication dose (perhaps a smart pill bottle pinged that the dose due at 8 AM was not taken by 9 AM), we send the nudge ASAP. Another example: if the patient’s wearable detects an abnormally high heart rate and the nudge is a calming exercise, deliver it at that moment.

    - If the nudge is not urgent, we schedule it for the next best time. Our function might consult the patient’s known routine or preferences. For instance, if the patient prefers evening messages, we schedule the educational tip for the evening. Or if our past data shows they are more responsive around lunchtime, we target that window. We might also enforce rules like “don’t send more than 2 nudges per day” to prevent overload.

    - We also consider the EAST Timely principle here – ensuring the prompt aligns with when the behavior should occur or when the user is most receptive. An example from our use case: scheduling a medication reminder exactly at the habitual time they take meds, or an exercise prompt when we know they’re usually free (not during work hours).

    The scheduling logic can become quite sophisticated. Initially, it might be simple (e.g., fixed times or immediate triggers), but over time it can learn. If the user repeatedly ignores morning reminders but responds in the afternoon, the system can shift future reminders to afternoon. In effect, the scheduling can adapt per user. The output of this step is a delivery_time (a timestamp or a trigger condition). Finally, when the time is right, the system dispatches the nudge through the chosen channel. For our prototype, this means simulating the send. In a real system, this would interface with an API (for push notifications, SMS gateway, email server, etc.). We also consider the Messenger effect here: the nudge can be sent by our health chatbot persona (“Your AF Coach”) or masked as coming from the doctor or clinic if that increases impact.

7. **Nudge Deployment:** This is the action of actually sending the message or executing the intervention. We have mostly covered it in step 6 – essentially, using the send_nudge function to deliver the content to the patient via the appropriate channel. It’s worth noting a few practical considerations:

- Channel and Modality: We ensure the infrastructure is in place for each channel (if via SMS, we need the patient’s phone number and a texting API; if via an app, the user should have push notifications enabled; if via email, ensure deliverability). In a Colab notebook context, we won’t actually send a text, but we might simulate it by printing the message or logging it.

- Logging the Attempt: Even as we deploy, we log that a nudge was sent, noting the time and content ID, to later correlate with outcomes.

- Fail-safes: If a message fails to send (e.g., network issues) or the channel is unavailable, the system might retry or use a backup channel. These details are part of a robust deployment but may be beyond our prototype scope.

8. **Feedback Loop – Tracking and Learning:** After the nudge is delivered, the cycle doesn’t end. We actively collect feedback to close the loop:

    - User Response: Did the patient respond to the nudge in any way? This could be explicit (e.g., clicking an “I did it” button, or replying to a chatbot) or implicit (their device later shows the pill bottle opened, indicating they took the dose). We record this response (or lack thereof).

    - Outcome: We check if the desired behavior happened. For instance, if it was a medication adherence nudge, we look at the medication log to see if the dose was eventually taken (perhaps within a reasonable window). If it was an exercise prompt, we might check step count data later in the day.

    - These data points (response and outcome) are logged with a timestamp in a database (or for our notebook, perhaps stored in a DataFrame). We update the patient’s profile based on this. For example, if the patient ignored multiple motivation nudges, we might update their motivation score (maybe it’s even lower, or we mark that approach as ineffective for them).

    - Over time, this feedback trains the system. We can analyze which types of nudges the patient responds to and which ones fall flat. This can inform adjustments: maybe the patient never clicks educational links we send – that might mean we should try a different content format or focus on another barrier first.

    - In a more advanced implementation, we would employ reinforcement learning (RL) here. The idea is to treat the problem as an RL scenario where the AI agent chooses an action (nudge type/content) and gets a reward based on the patient’s behavior (for example, +1 reward if the patient took their medication, 0 if not). The goal would be to maximize long-term adherence. A cluster-based RL approach could even be used, where the agent learns optimal strategies per patient cluster, as research has shown this can outperform one-size-fits-all policies. While full RL is beyond our current scope, laying the data groundwork with this feedback loop is essential for future learning.

# **Setting Up the Model w/ visualizations**

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('synth_AF_clusters_100k.csv')

# Define targets and drop outcome-based leakage columns
targets = ['macro_cluster', 'ttm_band', 'risk_tier']
drop_cols = ['patient_id', 'arch_prior', 'flags', 'subcluster_id', 'inr']
X = df.drop(columns=targets + drop_cols)

# Encode categorical variables
X['caregiver_support'] = X['caregiver_support'].astype('category')
X = pd.get_dummies(X, drop_first=True)

# Normalize continuous features
continuous_features = [
    'age', 'bmi', 'egfr', 'creatinine', 'la_size_mm', 'lvef',
    'cha2ds2_vasc', 'has_bled', 'motivation', 'risk_perc',
    'literacy_conf', 'autonomy', 'digital_comfort', 'open_rate2',
    'reply_rate2', 'latency_min2', 'ERS', 'pdc90', 'refill_streak',
    'visit_kept_rate', 'lab_on_time', 'BRI'
]
scaler = StandardScaler()
X[continuous_features] = scaler.fit_transform(X[continuous_features])

# Define a model training and evaluation function
def train_model(label):
    y = df[label]
    X_train, X_val, y_train, y_val = train_test_split(X, y, stratify=y, test_size=0.2, random_state=42)
    clf = RandomForestClassifier(n_estimators=100, random_state=0)
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_val)

    # Metrics
    print(f"\n=== {label.upper()} CLASSIFICATION RESULTS ===")
    print(f"Accuracy: {accuracy_score(y_val, y_pred):.4f}")
    print(classification_report(y_val, y_pred))

    # Confusion Matrix
    labels_sorted = sorted(y.unique())
    cm = confusion_matrix(y_val, y_pred, labels=labels_sorted)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels_sorted, yticklabels=labels_sorted)
    plt.title(f"{label.upper()} Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()
    return clf

# Run models
model_macro = train_model('macro_cluster')
model_ttm = train_model('ttm_band')
model_risk = train_model('risk_tier')
```
